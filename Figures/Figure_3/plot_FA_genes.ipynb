Figure 3e, g, i — ERG, RUNX1, MYB (same sc.pl.embedding call as the paper)


```python
from pathlib import Path
import warnings
import mudata
import muon
import scanpy as sc

warnings.filterwarnings("ignore", category=FutureWarning)
warnings.filterwarnings("ignore", message="Cannot join columns")
mudata.set_options(pull_on_update=False)

DATA = Path("PATH_TO_DATA")
sc.set_figure_params(dpi=100, fontsize=12, frameon=False, figsize=(6.4, 4.8))
newdata = muon.read(DATA / "TotalVI/New/PB2/TotalVI_PB2_TEAseq_CLEAN_TF.h5mu")
```

ERG


```python
sc.pl.embedding(
    newdata.mod['rna_subset'],
    basis="draw_graph_fa",
    color=["ERG"],
    frameon=False,
    ncols=4,
    color_map='plasma',
    size=100,
    title='',
    layer='denoised_rna',
    legend_fontsize=20, # adjust this value to change the size of the legend
    legend_loc='on data' if False else None, # this will remove the legend
    vmax=0.0035,
    vmin=0
)
```
